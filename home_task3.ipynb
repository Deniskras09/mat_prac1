```python
import numpy as np
import matplotlib.pyplot as plt
data = np.loadtxt('https://lectures.scientific-python.org/_downloads/087d8fe936fcb7a4762b9850b932cf9e/populations.txt')
years, hares, lynxes, carrots = data.T

x = years[:-1]
y = hares[:-1]

a, b = np.polyfit(x, y, deg = 1)

predict = a * 1920 + b
true_val = hares[-1]
error = abs(predict - true_val) / true_val

plt.figure(figsize = (15,8))
plt.plot(x, y, label = 'Данные до 1920', color = 'blue')
plt.plot(years, a * years + b, label = 'Лин. модель', color = 'orange')
plt.scatter(1920, true_val, color = 'olive', label = 'Истинное 1920')
plt.scatter(1920, predict, color = 'brown', label = 'Прогноз')
plt.ylabel('Популяция зайцев')
plt.xlabel('Год')
plt.xticks(years)
plt.yticks(np.arange(0, 80000, 5000))
plt.grid(True, linestyle = '--', color = 'pink')
plt.legend()
plt.show()
print("Относительная ошибка = ", error) 

```
